```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path(r"C:\dev\noo-far-pipeline")
sys.path.insert(0, str(PROJECT_ROOT))

from rag.retriever import retrieve
```

```python
# Fonction de comparaison

def comparer_modes(question, k=3):
    """Compare les 3 modes de récupération pour une question donnée."""
    print(f"{'='*70}")
    print(f"QUESTION : {question}")
    print(f"{'='*70}\n")

    for mode in ["semantic", "bm25", "hybrid"]:
        resultats = retrieve(question, mode=mode)
        print(f"--- Mode : {mode} ---")
        for i, (doc, meta) in enumerate(resultats, 1):
            source = meta["source"]
            apercu = doc[:80].replace("\n", " ")
            print(f"  {i}. {source} — {apercu}...")
        print()


def comparer_en_tableau(questions):
    """Tableau récapitulatif : top-1 par mode et par question."""
    lignes = []
    for q in questions:
        ligne = {"question": q}
        for mode in ["semantic", "bm25", "hybrid"]:
            resultats = retrieve(q, mode=mode)
            top1 = resultats[0][1]["source"] if resultats else None
            ligne[f"top1_{mode}"] = top1
        lignes.append(ligne)
    return pd.DataFrame(lignes)
```

```python
# Comparaion  sur les 5 questions

questions = [
    "Quand vacciner mes vaches ?",
    "Ma vache a de la fièvre, que faire ?",
    "Combien de fois traire par jour ?",
    "Quelle alimentation pour une vache en lactation ?",
    "Ma vache n'arrive pas à se reproduire",
]

for q in questions:
    comparer_modes(q)

```

```python
# Vue tabulaire, plus lisible pour comparer d'un coup d'œil
df_comparaison = comparer_en_tableau(questions)
df_comparaison
```
