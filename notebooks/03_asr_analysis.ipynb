## Section 1 — ASR Baseline

```python
MODELE = "whisper-medium"
csv_path = audio_dir / f"transcriptions_{MODELE}_fr.csv"

df_transcriptions = pd.read_csv(csv_path, encoding="utf-8-sig")
print(f"{len(df_transcriptions)} transcriptions chargées depuis {csv_path.name}")
df_transcriptions
```

```python
import matplotlib.pyplot as plt
import numpy as np
import librosa.display
from IPython.display import Audio, display

def observer_audio(audio_path, transcription):
    """Affiche forme d'onde + spectrogramme + player + transcription pour un audio."""
    audio, sr = librosa.load(str(audio_path), sr=16000)

    fig, axes = plt.subplots(2, 1, figsize=(12, 6))

    librosa.display.waveshow(audio, sr=sr, ax=axes[0])
    axes[0].set_title(f"{audio_path.name} — forme d'onde")

    D = librosa.stft(audio)
    S_db = librosa.amplitude_to_db(np.abs(D), ref=np.max)
    librosa.display.specshow(S_db, sr=sr, x_axis="time", y_axis="hz", ax=axes[1])
    axes[1].set_title("Spectrogramme")

    plt.tight_layout()
    plt.show()

    print(f"Transcription Whisper : {transcription}")
    display(Audio(audio, rate=sr))
```

```python
for _, row in df_transcriptions.iterrows():
    audio_path = audio_dir / row["fichier"]
    observer_audio(audio_path, row["transcription"])
```
